```
import pandas as pd

df = pd.read_csv(
    "enade2017_ciencia_computacao_agregado.csv",
    sep=";",
    encoding="utf-8-sig"
)

df.head()
```

```
print("Quantidade de linhas:", df.shape[0])
print("Quantidade de colunas:", df.shape[1])

print("\nValores ausentes em MEDIA_NT_GER:")
print(df["MEDIA_NT_GER"].isna().sum())

print("\nResumo da MEDIA_NT_GER:")
print(df["MEDIA_NT_GER"].describe())
```

```
df_modelo = df.dropna(subset=["MEDIA_NT_GER"]).copy()

print("Cursos usados:", df_modelo.shape[0])
```

```
df_modelo["CLASSE_DESEMPENHO"] = pd.qcut(
    df_modelo["MEDIA_NT_GER"],
    q=3,
    labels=["Baixo", "Médio", "Alto"]
)

print(df_modelo["CLASSE_DESEMPENHO"].value_counts())
```
